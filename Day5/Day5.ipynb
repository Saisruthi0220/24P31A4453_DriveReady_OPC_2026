```python
import os
import fitz
import numpy as np
import faiss

from sentence_transformers import SentenceTransformer
from openai import OpenAI

from google.colab import files, userdata
```

```python
uploaded = files.upload()

pdf_path = next(iter(uploaded.keys()))

print("Uploaded file:", pdf_path)
```

```python
doc = fitz.open(pdf_path)

documents = []

for page_number, page in enumerate(doc, start=1):

    text = page.get_text("text").strip()

    if text:
        documents.append({
            "page": page_number,
            "text": text
        })

print("Total pages extracted:", len(documents))

for item in documents:
    print(f"\n--- PAGE {item['page']} ---")
    print(item["text"])
```

```python
total_characters = sum(
    len(item["text"]) for item in documents
)

print("Total pages:", len(documents))
print("Total characters:", total_characters)

if total_characters > 0:
    print("PDF extraction successful!")
else:
    print("No text found in PDF.")
```

```python
chunk_size = 500
overlap = 100

chunks = []

for item in documents:

    words = item["text"].split()

    start = 0

    while start < len(words):

        end = start + chunk_size

        chunk_text = " ".join(words[start:end])

        if chunk_text.strip():
            chunks.append({
                "text": chunk_text,
                "page": item["page"]
            })

        start += chunk_size - overlap

print("Total chunks created:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1} | Page {chunk['page']}")
    print(chunk["text"][:200])
```

```python
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully!")
```

```python
texts = [chunk["text"] for chunk in chunks]

embeddings = embedding_model.encode(
    texts,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

embeddings = embeddings.astype("float32")

print("Embedding shape:", embeddings.shape)
print("Embeddings generated successfully!")
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(embeddings)

print("FAISS index created successfully!")
print("Total vectors stored:", index.ntotal)
```

```python
def retrieve_documents(question, top_k=4):

    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    scores, indices = index.search(
        query_embedding,
        min(top_k, len(chunks))
    )

    results = []

    for score, idx in zip(scores[0], indices[0]):

        if idx == -1:
            continue

        results.append({
            "text": chunks[idx]["text"],
            "page": chunks[idx]["page"],
            "score": float(score)
        })

    return results
```

```python
question = "What are the library opening hours?"

results = retrieve_documents(question)

for result in results:

    print("Page:", result["page"])
    print("Similarity:", round(result["score"], 3))
    print("Content:", result["text"])
    print("-" * 50)
```

```python
def ask_rag(question):

    results = retrieve_documents(question, top_k=4)

    if not results:
        return "No relevant information found in the PDF."

    context = ""

    for i, result in enumerate(results):
        context += f"""
        Context {i+1}
        Page Number: {result['page']}
        Content: {result['text']}
        """

    system_prompt = """
    You are a helpful PDF question-answering assistant.

    Answer ONLY using the provided PDF context.

    Rules:
    1. Do not use outside knowledge.
    2. Do not invent facts.
    3. If information is unavailable, say:
       Information not available in the uploaded document.
    4. Provide clear answers.
    5. Mention source page numbers.
    """

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {"role": "system", "content": system_prompt},
            {
                "role": "user",
                "content": f"""
                PDF CONTEXT:
                {context}

                QUESTION:
                {question}
                """
            }
        ],
        temperature=0.2
    )

    return response.choices[0].message.content
```

```python
question = "How many books can a student borrow?"

answer = ask_rag(question)

print("Question:", question)
print("\nAnswer:", answer)
```

```python
test_questions = [
    "How many books can a student borrow?",
    "What are the library opening hours?",
    "What is the late return fee?",
    "How many study rooms are available?",
    "What is the library support email?"
]

for question in test_questions:
    print("\nQuestion:", question)
    print("Answer:", ask_rag(question))
    print("-" * 60)
```
